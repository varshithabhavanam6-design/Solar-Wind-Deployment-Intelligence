```python
import pandas as pd
import numpy as np

# List of files to process
files = ["West zone.xlsx", "Central zone.xlsx", "South zone.xlsx", "East zone.xlsx", "North zone.xlsx"]

for file in files:
    # Load the Excel file
    df = pd.read_excel(file, sheet_name="Sheet1")

    # Identify rows where LAT.1 (column I, index 8) ends with 0.5
    mask = df.iloc[:, 8] % 1 == 0.5  # Column I is index 8 (0-based)

    # Columns I to M (9th to 13th column, index 8 to 12)
    cols_to_filter = df.columns[8:13]

    # Create a copy of the dataframe but only modify columns after the 8th
    df_filtered = df.copy()

    # Apply NaN only for columns I to M, without changing the first 8 columns
    df_filtered.loc[~mask, cols_to_filter] = np.nan

    # **Shift non-NaN values UP in columns I to M**
    for col in cols_to_filter:
        df_filtered[col] = df_filtered[col].dropna().reset_index(drop=True)

    # Save the filtered data with a new name
    output_file = f"Final_{file}"
    df_filtered.to_excel(output_file, index=False)

    print(f"Filtered file saved as '{output_file}'")
```
